# Customer Churn Model Experiments

This notebook trains and compares two churn models on the customer-level feature set: logistic regression and random forest. The final target is to pick the model that best balances recall and precision for a real business-risk scenario.

```python
import pandas as pd
from src.analytics import load_business_data
from src.feature_engineering import build_customer_features
from src.ml_model import ChurnModelPipeline, export_model_metadata
```
