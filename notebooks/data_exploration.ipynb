# Dataset Exploration and Business Analytics

This notebook explores the synthetic e-commerce data created for the business intelligence system. It reviews data quality, trends, category performance, customer spending, and churn indicators.

```python
import pandas as pd
import numpy as np
from pathlib import Path

from src.analytics import (
    load_business_data,
    total_revenue,
    monthly_revenue,
    category_performance,
    product_performance,
    city_performance,
    customer_spending,
    customer_segmentation,
    churn_statistics,
)

data = load_business_data()
customers = data['customers']
products = data['products']
orders = data['orders']
order_items = data['order_items']

customers.head()
```

```python
# Missing values by table
for name, df in data.items():
    print(f'{name}: missing={int(df.isna().sum().sum())}')
    print(df.isna().sum().sort_values(ascending=False).head())
    print('---')
```

```python
# Basic row counts
for name, df in data.items():
    print(f'{name}: {len(df)} rows')
```

```python
# Distribution analysis
customers[['age', 'tenure_months', 'monthly_spend', 'support_calls', 'purchase_frequency']].describe().round(2)
```

```python
# Monthly revenue trend
monthly_rev = monthly_revenue(orders)
monthly_rev.head()
```

```python
# Revenue trend view
monthly_rev['total_revenue'].plot(kind='line', figsize=(12,4), title='Monthly Revenue Trend')
```

```python
# Category analysis
category_perf = category_performance(products, order_items)
category_perf.head(10)
```

```python
# Top products
top_products = product_performance(products, order_items).head(10)
top_products
```

```python
# City performance
city_perf = city_performance(customers, orders)
city_perf.head(10)
```

```python
# Customer spending distribution
customer_spends = customer_spending(customers, orders)
customer_spends[['customer_id', 'total_spent', 'total_orders']].head(10)
```

```python
# Customer segmentation
segments = customer_segmentation(customers, orders)
segments['segment'].value_counts()
```

```python
# Churn analysis
churn_details = churn_statistics(customers)
churn_details
```

```python
# Churn rate by city
customers.groupby('city')['churn'].mean().mul(100).round(2).sort_values(ascending=False).head(10)
```

```python
# Business summary
print('Total Revenue:', total_revenue(orders_df=orders, order_items_df=order_items))
print('Total Orders:', len(orders))
print('Total Customers:', len(customers))
print('Avg Order Value:', round(total_revenue(orders_df=orders, order_items_df=order_items) / len(orders), 2))
```
